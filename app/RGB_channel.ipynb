```python
import cv2
import matplotlib.pyplot as plt

# 1. Define the path safely!
# The 'r' before the string tells Python to treat backslashes normally
image_path = r"captures\2026-05-05-17-55-33\3.png"

# 2. Read the image (loads as BGR)
img = cv2.imread(image_path)

# 3. The Safety Net: Check if the image actually loaded
if img is None:
    print(f"ERROR: OpenCV could not find the image at: {image_path}")
    print("Check your folder structure and ensure you are running the script in the correct directory.")
else:
    print("Image loaded successfully!")
    
    # 4. Convert BGR to RGB
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    # 5. Display the image
    # We use Matplotlib here because cv2.imshow() expects BGR format,
    # but Matplotlib expects the RGB format we just created!
    plt.figure(figsize=(8, 6))
    plt.imshow(img_rgb)
    plt.title("Converted RGB Image")
    plt.axis('off') # Hides the axis numbers
    plt.show()
    
    # Optional: If you want to save the new image as a file
    # (Note: we have to convert it BACK to BGR to save it with cv2)
    # img_bgr_to_save = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)
    # cv2.imwrite("fixed_image.png", img_bgr_to_save)
```
