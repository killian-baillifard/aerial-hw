## fix rbg - bgr for images from day 1


```python


import cv2
import os

def fix_image_channels(input_folder, output_folder):
    """
    Scans an input folder for images, fixes the RGB/BGR color channel swap,
    and saves the corrected images to an output folder.
    """
    
    # 1. Safety Check: Does the input folder even exist?
    if not os.path.exists(input_folder):
        print(f"ERROR: The input folder '{input_folder}' does not exist.")
        return 0, 0
        
    # 2. Create the output folder (if it doesn't exist)
    os.makedirs(output_folder, exist_ok=True)

    processed_count = 0
    failed_count = 0

    print(f"Scanning folder: {input_folder}...")
    
    # 3. Loop through every file
    for filename in os.listdir(input_folder):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
            
            image_path = os.path.join(input_folder, filename)
            new_image_path = os.path.join(output_folder, filename)
            
            img = cv2.imread(image_path)
            
            if img is None:
                print(f"  [!] Failed to read: {filename}")
                failed_count += 1
            else:
                # 4. Convert and Save
                img_bgr = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)
                success = cv2.imwrite(new_image_path, img_bgr)
                
                if success:
                    print(f"  [+] Fixed and saved: {filename}")
                    processed_count += 1
                else:
                    print(f"  [!] Failed to save: {filename}")
                    failed_count += 1

    # 5. Print Summary
    print("-" * 30)
    print(f"Batch Processing Complete for: {os.path.basename(input_folder)}")
    print(f"Successfully fixed: {processed_count} images")
    if failed_count > 0:
        print(f"Failed to process: {failed_count} images")
        
    # Return the numbers in case your main script wants to use them
    return processed_count, failed_count
```

## Combine the data into 1 folder

```python
import os
import shutil
import numpy as np

def combine_datasets(input_folders, output_folder):
    """
    Combines multiple image folders and their respective measures.npy files 
    into one master directory.
    """
    # create the master output folder
    os.makedirs(output_folder, exist_ok=True)

    all_measures = []
    total_images_copied = 0

    print("Starting consolidation process...\n")

    # Iterate through the folders
    for folder in input_folders:
        if not os.path.exists(folder):
            print(f"[!] Warning: Folder not found, skipping: {folder}")
            continue

        # Extract just the folder name
        folder_name = os.path.basename(os.path.normpath(folder))
        npy_path = os.path.join(folder, 'measures.npy')

        # Check if the .npy file exists
        if not os.path.exists(npy_path):
            print(f"[!] Warning: measures.npy not found in {folder_name}, skipping.")
            continue

        # Load the NumPy array and store it list
        measures = np.load(npy_path)
        all_measures.append(measures)

        # 4. Find all images and sort them NUMERICALLY (0.png, 1.png, ..., 10.png)
        # This ensures they map perfectly to the rows in the .npy file
        images = [f for f in os.listdir(folder) if f.lower().endswith(('.png', '.jpg'))]
        images.sort(key=lambda x: int(os.path.splitext(x)[0]))

        # Safety Check: number of images = number of .npy rows?
        if len(images) != len(measures):
            print(f"  [!] DANGER: Mismatch in {folder_name}! {len(images)} images vs {len(measures)} data rows.")

        # Copy and rename the images
        for i, img_name in enumerate(images):
            src_img = os.path.join(folder, img_name)
            # Zero-pad the index to 4 digits to guarantee alphabetical sorting works later
            # Result: 2026-05-05-18-00-31_0000.png
            new_name = f"{folder_name}_{i:04d}.png"
            dst_img = os.path.join(output_folder, new_name)
            
            # shutil.copy2 copies the file and preserves creation/modification metadata
            shutil.copy2(src_img, dst_img)

        total_images_copied += len(images)
        print(f"[+] Processed {folder_name} (Copied {len(images)} images)")

    # 6. Combine all the .npy arrays into one massive array
    if all_measures:
        # Concatenate along the first axis (rows)
        combined_npy = np.concatenate(all_measures, axis=0)
        out_npy_path = os.path.join(output_folder, 'combined_measures.npy')
        
        # Save the master .npy file
        np.save(out_npy_path, combined_npy)
        
        print("\n" + "=" * 40)
        print("Consolidation Complete!")
        print(f"Total images copied: {total_images_copied}")
        print(f"Combined NPY shape:  {combined_npy.shape}")
        print(f"Output Location:     {output_folder}")
    else:
        print("\n[!] No valid data found to combine. Check your paths.")
```

```python
# Specify the exact paths of the folders you want to combine
folders_to_combine = [
    #pictures from day1
    r"..\saved_captures_fixed\2026-05-05-17-55-33_fixed",
    r"..\saved_captures_fixed\2026-05-05-17-58-20_fixed",
    r"..\saved_captures_fixed\2026-05-05-18-00-31_fixed",
    r"..\saved_captures_fixed\2026-05-05-18-05-02_fixed",
    r"..\saved_captures_fixed\2026-05-05-18-57-43_fixed",

    #pictures from day2
    r"..\saved_captures_fixed\2026-05-06-16-28-06_fixed",
    r"..\saved_captures_fixed\2026-05-06-16-43-10_fixed",
    r"..\saved_captures_fixed\2026-05-06-16-46-08_fixed",
    r"..\saved_captures_fixed\2026-05-06-16-50-00_fixed",
]

# Specify where the massive combined folder should go
master_output_directory = r"..\saved_captures\MASTER_DATASET"

combine_datasets(folders_to_combine, master_output_directory)
```
